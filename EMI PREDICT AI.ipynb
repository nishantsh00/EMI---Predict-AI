```python
def clean_gender(x):
    x = str(x).strip().lower()
    if x in ['male', 'm']:
        return 'male'
    elif x in ['female', 'f']:
        return 'female'
    else:
        return 'other'  # Optional: treat unexplained categories

df['gender'] = df['gender'].apply(clean_gender)
```

```python
df
```

```python
df["education"].value_counts()
```

```python
allowed_categories = ["Graduate", "Post Graduate", "High School", "Professional"]

df['education'] = df['education'].str.strip().str.title()

df['education'] = df['education'].apply(lambda x: x if x in allowed_categories else "Others")
```

```python
df["education"].fillna(df["education"].mode, inplace= True)
```

```python
from sklearn.impute import SimpleImputer
```

```python
# Convert columns to numeric 
num_cols = ["age","monthly_salary","years_of_employment","monthly_rent","school_fees","travel_expenses","other_monthly_expenses","current_emi_amount","groceries_utilities","college_fees","bank_balance", "credit_score","emergency_fund","max_monthly_emi"]
df[num_cols] = df[num_cols].apply(pd.to_numeric, errors="coerce")
```

```python
imp = SimpleImputer(strategy="median")
df[["age","monthly_salary","monthly_rent","credit_score","bank_balance","emergency_fund"]] = imp.fit_transform(df[["age","monthly_salary","monthly_rent","credit_score","bank_balance","emergency_fund"]])
```

## FEATURE ENGINEERING

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler, LabelEncoder
```

```python
df.columns
```

```python
df[["gender", "marital_status", "education", "employment_type", "company_type", "house_type","existing_loans", "emi_scenario","emi_eligibility"]].head()
```

```python
cat_cols = ["gender", "marital_status", "education", "employment_type", "company_type", "house_type","existing_loans", "emi_scenario"]
```

```python
# CONVERTING TO STRING
df[cat_cols] = df[cat_cols].astype(str)
```

```python
# ENCODING COLUMNS
oc = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
encoded_cols = oc.fit_transform(df[cat_cols])
```

```python
encoded_col_names = oc.get_feature_names_out(cat_cols)
df_encoded = pd.DataFrame(encoded_cols, columns=encoded_col_names, index=df.index)
df_encoded.head()
```

```python
# ENCODE THE LABEL COLUMN
lc = LabelEncoder()
df["emi_eligibility"]  = lc.fit_transform(df["emi_eligibility"])
```

```python
# COLUMNS FOR SCALING
scaled_cols = df.drop(columns=cat_cols + ["emi_eligibility", "max_monthly_emi"])
```

```python
scaled_cols.head()
```

```python
# SCALING NUM COLUMNS
st = StandardScaler()
scaled_data = st.fit_transform(scaled_cols)
df_scaled = pd.DataFrame(scaled_data, columns=scaled_cols.columns)
df_scaled.info()
```

```python
df_scaled.head()
```

# REGRESSION MODEL 

```python
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from xgboost import XGBRegressor, XGBClassifier
from sklearn.metrics import r2_score, mean_absolute_error, root_mean_squared_error, mean_absolute_percentage_error
from sklearn.metrics import accuracy_score, f1_score, recall_score , roc_auc_score, precision_score
import pickle
```

```python
X_r = pd.concat([df_scaled, df_encoded], axis=1)
```

```python
y_r = df["max_monthly_emi"].copy()
```

```python
# TRAIN TEST SPLIT
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_r_train, X_r_test, y_r_train,y_r_test = train_test_split(X_r, y_r, test_size=0.2, random_state=42)
```

```python
print(X_r_train.shape, y_r_train.shape)
print(X_r_test.shape, y_r_test.shape)
```

```python
models = {
    "LinearRegression": LinearRegression(),
    "RandomForestRegressor": RandomForestRegressor(n_estimators=100),
    "XGBoostRegressor": XGBRegressor(),
    "DecisionTreeRegressor": DecisionTreeRegressor()
}

# Train and evaluate each model
for name, model in models.items():
    model.fit(X_r_train, y_r_train)               # Train model
    y_pred = model.predict(X_r_test)              # Predict test set
    rmse = root_mean_squared_error(y_r_test, y_pred)
    mae = mean_absolute_error(y_r_test, y_pred)
    r2 = r2_score(y_r_test, y_pred)
    print(f"{name} - RMSE: {rmse:.4f}, MAE: {mae:.4f}, R2: {r2:.4f}")
```

```python
# Train final regression model
xgb_reg = XGBRegressor()
xgb_reg.fit(X_r_train, y_r_train)

# Save regression model to file
with open('xgboost_regression_model.pkl', 'wb') as f:
    pickle.dump(xgb_reg, f)

print("XGBoost REG models saved with pickle.")
```

# CLASSIFICATION MODEL

```python
X_c = pd.concat([df_scaled, df_encoded], axis=1)
```

```python
y_c = df["emi_eligibility"].copy()
```

```python
X_c_train, X_c_test, y_c_train,y_c_test = train_test_split(X_c, y_c, test_size=0.2, random_state=42)
```

```python
print(X_c_train.shape, y_c_train.shape)
print(X_c_test.shape, y_c_test.shape)
```

```python
class_models = {
    "LogisticRegression": LogisticRegression(max_iter=1000),
    "RandomForestClassifier": RandomForestClassifier(n_estimators=100),
    "XGBoostClassifier": XGBClassifier(use_label_encoder=False, eval_metric='logloss'),
    "DecisionTreeClassifier": DecisionTreeClassifier()
}

for name, model in class_models.items():
    model.fit(X_c_train, y_c_train)                      # Train
    y_pred = model.predict(X_c_test)                     # Predict
    
    
    acc = accuracy_score(y_c_test, y_pred)
    prec = precision_score(y_c_test, y_pred, average='weighted')
    rec = recall_score(y_c_test, y_pred, average='weighted')
    f1 = f1_score(y_c_test, y_pred, average='weighted')
    
    output = f"{name} - ACC: {acc:.4f}, Prec: {prec:.4f}, Rec: {rec:.4f}, F1: {f1:.4f}"
    
    print(output)
```

```python
# Train final classification model
xgb_clf = XGBClassifier(use_label_encoder=False, eval_metric='logloss')
xgb_clf.fit(X_c_train, y_c_train)

# Save classification model to file
with open('xgboost_classification_model.pkl', 'wb') as f:
    pickle.dump(xgb_clf, f)

print("XGBoost CLASS models saved with pickle.")
```
